# Slicing

# Normalize

```python
mean_a = np.mean(a)
mean_a
```

```python
std_a = np.std(a)
std_a
```

```python
norm_a = (a - mean_a) / std_a
norm_a
```

```python
np.mean(norm_a)
```

```python
np.std(norm_a)
```
